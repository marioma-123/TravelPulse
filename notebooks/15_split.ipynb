```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import GroupShuffleSplit
```

```python
PROC = Path("../data/processed")
df = pd.read_csv(PROC / "listings_final.csv", low_memory=False)
```

```python
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(df, groups=df["host_id"]))
```

```python
df["split"] = "train"
df.loc[df.index[test_idx], "split"] = "test"
```

```python
print(df["split"].value_counts())
print("hosts in both:", len(set(df.loc[df.split == "train", "host_id"]) & set(df.loc[df.split == "test", "host_id"])))
```

```python
df[["city", "id", "split"]].to_csv(PROC / "split_assignment.csv", index=False)
```
