```python
import os
import pandas as pd
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
```

```python
cities = ["Amsterdam", "Barcelona", "Berlin", "Columbus"]
short_cols = {"id", "listing_id", "date", "available", "host_id"}

for name in ["listings", "calendar"]:
    sets = {}
    for city in cities:
        try:
            sets[city] = list(pd.read_csv(f"../data/raw/{city}_{name}.csv", nrows=0).columns)
        except Exception as e:
            print(f"Error in {city}_{name}: {e}")
            
    base = sets["Berlin"]
    for city, cols in sets.items():
        if cols != base:
            print(f"{city}_{name} columns differ:", set(cols) ^ set(base))
            
    ddl = ",\n    ".join(
        f"`{c}` {'VARCHAR(50)' if c in short_cols else 'LONGTEXT'}" for c in base
    )
    print(f"\nCREATE TABLE raw_{name} (\n    city VARCHAR(50),\n    {ddl}\n);\n")
```

```python
load_dotenv()
engine = create_engine(
    f"mysql+pymysql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}"
    f"@{os.getenv('DB_HOST')}:{os.getenv('DB_PORT')}/{os.getenv('DB_NAME')}?charset=utf8mb4"
)

with engine.connect() as conn:
    print(conn.execute(text("SELECT DATABASE()")).scalar())
```

```python
def load_csv(path, city, table):
    with engine.begin() as conn:
        conn.execute(text(f"DELETE FROM {table} WHERE city = :c"), {"c": city})
    total = 0
    for chunk in pd.read_csv(path, dtype=str, chunksize=50_000):
        chunk.insert(0, "city", city)
        chunk = chunk.astype(object).where(chunk.notna(), None)
        chunk.to_sql(table, engine, if_exists="append", index=False,chunksize=10_000, method="multi")
        total += len(chunk)
        print(f"  {city} {table}: {total:,}")
    print(f"{city} → {table}: {total:,} صف")
```

```python
#Columbus
load_csv("../data/raw/Columbus_listings.csv", "Columbus", "raw_listings")
load_csv("../data/raw/Columbus_calendar.csv", "Columbus", "raw_calendar")
```
